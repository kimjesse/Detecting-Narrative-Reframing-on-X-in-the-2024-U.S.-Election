```python
import pandas as pd
from pathlib import Path
```

```python
directory = Path("../fixed")
files = list(directory.glob("*may_july*.csv"))
```

```python
df = [pd.read_csv(file, low_memory=False) for file in files]
```

```python
df_concat = pd.concat(df,ignore_index=True,axis=0)
```

```python
strat = df_concat.groupby('week', group_keys=False).apply(lambda x: x.sample(n=min(len(x),500), random_state=26))
```
